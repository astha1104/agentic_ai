```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-3-flash-preview")

response=llm.invoke([
    ["system","you are a helpful assitant that answers in short one line."],
    ["human","How many moons does jupiter have?"]
])
print(response.text)                          
```

```python
prompt="write a single tagline for a quicky shoes shop.only output the tagline."
llm=ChatGoogleGenerativeAI(model="gemini-3-flash-preview",temperature=0)

response=llm.invoke(prompt)
print(response.text)                          
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-3-flash-preview",temperature=1)

response=llm.invoke(prompt)
print(response.text) 
```
